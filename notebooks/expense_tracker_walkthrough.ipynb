## Phase 7 - Charts (inline)

```
import seaborn as sns
cs = an.category_summary(df)
plt.figure(figsize=(9,5))
sns.barplot(data=cs, x='Total', y='Category', hue='Category', legend=False)
plt.title('Total Expense by Category')
plt.show()
```

```
ms = an.monthly_summary(df)
ms.plot(x='Month', y=['Income','Expense','Savings'], marker='o', figsize=(10,4), title='Monthly trend')
plt.xticks(rotation=45)
plt.show()
```

```
pv = an.category_month_matrix(df)
plt.figure(figsize=(12,5))
sns.heatmap(pv, cmap='YlOrRd', annot=True, fmt='.0f')
plt.title('Category x Month')
plt.show()
```
